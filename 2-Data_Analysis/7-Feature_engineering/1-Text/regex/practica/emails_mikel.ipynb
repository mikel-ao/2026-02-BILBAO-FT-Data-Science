# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python
# Explicación del patrón r'@([\w.-]+\.[a-z]{2,})':
# @          -> Busca el símbolo literal @
# (          -> Inicia el grupo de captura (lo que queremos extraer)
#  [\w.-]+   -> Busca uno o más caracteres (letras, números, puntos o guiones)
#  \.        -> Busca un punto literal antes de la extensión
#  [a-z]{2,} -> Busca la extensión (mínimo 2 letras, como .es, .com, .tech)
# )          -> Cierra el grupo de captura

# 1. Compilamos el patrón
# re.MULTILINE permite que ^ y $ funcionen al principio y final de cada línea
# Usamos un grupo de captura () para extraer solo lo que está después del @
patron = re.compile(r'@([\w.-]+\.[a-z]{2,})', re.MULTILINE)

# 2. Usamos finditer para obtener un iterador de objetos "match"
# Esto es más eficiente en memoria que findall para textos muy largos
matches = patron.finditer(emails)

# 3. Extraemos el contenido del grupo 1 de cada coincidencia
dominios = [match.group(1) for match in matches]
for d in dominios:
    print(d)
```
